```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#Dirty Values - NA/NULL/Blank/Spaces 

df.replace(
    ['N/A','NULL', '',' '],np.nan,inplace=True
)
```

```python
df.isnull().sum()
```

```python
# Extra Spaces Remove

df=df.apply(lambda x:x.str.strip() if x.dtype == "str" else x)
df.info()
```

```python
#Standardize Text

df["Country"] = df["Country"].str.title()

df["Genre"] = (df["Genre"].str.title())

df["Subscription_Plan"] = (df["Subscription_Plan"].str.title())

df["Payment_Method"] = (df["Payment_Method"].str.title())

df["Account_Status"] = (df["Account_Status"].str.title())
```

```python
#Counting elements

df["Country"].value_counts()
```

```python
 #Numeric columns

numeric_cols=['Age','Watch_Hours','Monthly_Fee','Movies_Watched','Episodes_Watched','Avg_Rating','Monthly_Sessions', 'Download_Count']
```

```python
for col in numeric_cols:
    df[col]=pd.to_numeric(df[col],errors ='coerce')
```

```python
df.describe()
```

```python
df=df[(df['Age']>=18) & (df['Age']<=100)]
```

```python
df['Age'].describe()
```

```python
#Negative Values

df[df['Watch_Hours'] >= 0]
df[df['Monthly_Fee'] >= 0]
```

```python
df['Watch_Hours'] = df['Watch_Hours'].abs()

df.loc[df['Watch_Hours'] < 0, 'Watch_Hours'] = 0

df = df[df['Watch_Hours'] >= 0]

(df['Watch_Hours'] < 0).sum()

df[df['Watch_Hours'] < 0]['Watch_Hours']
```

```python
#Data Cleaning (date)

df['Join_Date'] =pd.to_datetime(df['Join_Date'],errors='coerce')
df['Last_Watch_Date'] =pd.to_datetime(df['Last_Watch_Date'],errors='coerce')
```

```python
#Missing Values

df['Age']=df['Age'].fillna(
    df['Age'].mean()
)

df['Watch_Hours']=df['Watch_Hours'].fillna(
    df['Watch_Hours'].mean()
)

df['Monthly_Fee']=df['Monthly_Fee'].fillna(
    df['Monthly_Fee'].mean()
)
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna("Unknown")
df['Device'] = df['Device'].fillna("Unknown")
df['Genre'] = df['Genre'].fillna("Unknown")
```

```python
#Feature Engineering

#Customer Revenue

df['Customer_Revenue'] = df['Monthly_Fee'] *df['Monthly_Sessions']
```

```python
#Engagement Score

df['Engagement_Score'] =(df['Watch_Hours'] + df['Monthly_Sessions'] + df['Download_Count'])
```

```python
df[df['Watch_Hours'] >= 0]
```

```python
# Watch Category
df['Watch_Category'] = np.select(
    [
        df['Watch_Hours'] == 0,
        df['Watch_Hours'].between(1, 50),
        df['Watch_Hours'].between(50, 100),
        df['Watch_Hours'] > 100
    ],
    [
        'No Watch',
        'Low',
        'Medium',
        'High'
    ],
    default='Unknown'
)
```

```python
df.head()
```

```python
df['Watch_Category'].value_counts().sum()
```

```python
#Customer Age Group

bins =[17,25,35,50,100]
label =['18-25','26-35','36-50','51+']

df['Age_Group'] =pd.cut(df['Age'],bins =bins ,labels=label)
```

```python
df['Age_Group'].value_counts()
```

```python
#Account Active Flag

df['Active_Flag'] =np.where(df["Account_Status"]=="Active",1,0)
```

```python
df.info()
```

```python
# Define the columns you want to convert
columns_to_convert = [
    'Customer_ID', 'Customer_Name', 'Title', 'Content_Type', 
    'Genre', 'Country', 'Subscription_Plan', 'Device', 
    'Payment_Method', 'Account_Status'
]

# Create a dictionary mapping each column to the 'object' datatype
conversion_dict = {col: object for col in columns_to_convert}

# Convert the datatypes in your DataFrame
df = df.astype(conversion_dict)

# Verify the changes
print(df.dtypes)
```
